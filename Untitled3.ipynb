```python
import pandas as pd
import numpy as np
from prophet import Prophet
from sklearn.metrics import mean_absolute_percentage_error
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import warnings
warnings.filterwarnings("ignore")
import random

random.seed(42)
np.random.seed(42)

# ── Gerar dados simulados de fundição ─────────────────────────────────────────
from datetime import date, timedelta

def gerar_contas_receber(n=500):
    clientes = [f"MONTADORA-{i:02d}" for i in range(1, 8)]  # Ford, VW, Stellantis...
    categorias = ["Peças Fundidas", "Componentes Motor", "Chassi", "Suspensão"]
    status_list = ["Em aberto", "Pago", "Vencido", "Renegociado"]
    hoje = date.today()
    registros = []
    for i in range(n):
        emissao   = hoje - timedelta(days=random.randint(0, 90))
        vencimento = emissao + timedelta(days=random.randint(60, 120))  # prazo longo: montadoras
        valor     = round(random.uniform(10000, 300000), 2)
        status    = random.choices(status_list, weights=[0.4, 0.35, 0.15, 0.10])[0]
        registros.append({
            "id_titulo": f"AR-{i+1:06d}",
            "cliente":   random.choice(clientes),
            "categoria": random.choice(categorias),
            "valor_nominal":  valor,
            "valor_desconto": round(valor * random.uniform(0, 0.03), 2),
            "data_emissao":   emissao,
            "data_vencimento": vencimento,
            "status": status,
        })
    return pd.DataFrame(registros)

def gerar_contas_pagar(n=400):
    # Categorias reais de fundição
    fornecedores = {
        "Energia Elétrica": (50000, 200000),   # maior custo
        "Sucata/Ferro-gusa": (30000, 150000),  # matéria-prima
        "Folha de Pagamento": (80000, 120000), # salários
        "Impostos":          (20000, 80000),
        "Manutenção Fornos": (5000, 40000),
        "Gases Industriais": (3000, 20000),
    }
    status_list = ["Em aberto", "Pago", "Vencido", "Agendado"]
    hoje = date.today()
    registros = []
    for i in range(n):
        cat = random.choice(list(fornecedores.keys()))
        vmin, vmax = fornecedores[cat]
        emissao    = hoje - timedelta(days=random.randint(0, 60))
        # Energia vence dia 10, folha dia 5, outros variável
        if cat == "Folha de Pagamento":
            prox_mes = (emissao.replace(day=1) + timedelta(days=32)).replace(day=5)
            vencimento = prox_mes
        elif cat == "Energia Elétrica":
            prox_mes = (emissao.replace(day=1) + timedelta(days=32)).replace(day=10)
            vencimento = prox_mes
        else:
            vencimento = emissao + timedelta(days=random.randint(7, 30))
        valor  = round(random.uniform(vmin, vmax), 2)
        status = random.choices(status_list, weights=[0.3, 0.40, 0.10, 0.20])[0]
        registros.append({
            "id_titulo":      f"AP-{i+1:06d}",
            "categoria":      cat,
            "valor_nominal":  valor,
            "valor_desconto": round(valor * random.uniform(0, 0.01), 2),
            "data_emissao":   emissao,
            "data_vencimento": vencimento,
            "status": status,
            "prioridade": "Alta" if cat in ["Energia Elétrica", "Folha de Pagamento", "Impostos"] else "Normal",
        })
    return pd.DataFrame(registros)

def gerar_extrato(n=360):
    hoje = date.today()
    saldo = 800000.0
    registros = []
    for i in range(n):
        data = hoje - timedelta(days=n - i)
        # Padrão de fundição: recebe de montadoras em lotes grandes, paga todo dia
        dia = data.day
        if dia in [5, 6]:       # folha
            valor, tipo = round(random.uniform(80000, 120000), 2), "Débito"
        elif dia in [10, 11]:   # energia
            valor, tipo = round(random.uniform(50000, 180000), 2), "Débito"
        elif dia in [25, 26, 27]:  # recebimento de montadoras
            valor, tipo = round(random.uniform(100000, 400000), 2), "Crédito"
        else:
            valor = round(random.uniform(1000, 30000), 2)
            tipo  = random.choices(["Crédito", "Débito"], weights=[0.4, 0.6])[0]
        saldo += valor if tipo == "Crédito" else -valor
        registros.append({"data": data, "tipo": tipo, "valor": valor,
                          "valor_sinalizado": valor if tipo == "Crédito" else -valor,
                          "saldo": round(saldo, 2)})
    return pd.DataFrame(registros)

ar  = gerar_contas_receber()
ap  = gerar_contas_pagar()
ext = gerar_extrato()

print(f"✅ AR gerado:      {len(ar)} títulos  | Total: R$ {ar['valor_nominal'].sum():,.0f}")
print(f"✅ AP gerado:      {len(ap)} títulos  | Total: R$ {ap['valor_nominal'].sum():,.0f}")
print(f"✅ Extrato gerado: {len(ext)} dias    | Saldo final: R$ {ext['saldo'].iloc[-1]:,.0f}")
```

```python
# ── Silver: limpeza e classificação ──────────────────────────────────────────
hoje = date.today()

ar["valor_liquido"]    = ar["valor_nominal"] - ar["valor_desconto"]
ar["dias_vencimento"]  = pd.to_timedelta(ar["data_vencimento"] - hoje).dt.days
ar["situacao"]         = ar.apply(lambda r:
    "Pago" if r["status"] == "Pago"
    else "Vencido"       if r["dias_vencimento"] < 0
    else "Vence Hoje"    if r["dias_vencimento"] == 0
    else "Vence em 7d"   if r["dias_vencimento"] <= 7
    else "Vence em 30d"  if r["dias_vencimento"] <= 30
    else "Vence em +30d", axis=1)
ar["fl_inadimplente"] = (ar["status"] != "Pago") & (ar["dias_vencimento"] < -15)

ap["valor_liquido"]   = ap["valor_nominal"] - ap["valor_desconto"]
ap["dias_vencimento"] = pd.to_timedelta(ap["data_vencimento"] - hoje).dt.days
ap["situacao"]        = ap.apply(lambda r:
    "Pago" if r["status"] == "Pago"
    else "Vencido"      if r["dias_vencimento"] < 0
    else "Vence em 7d"  if r["dias_vencimento"] <= 7
    else "Vence em 30d" if r["dias_vencimento"] <= 30
    else "Vence em +30d", axis=1)

print("✅ Silver — AR")
print(ar["situacao"].value_counts().to_string())
print("\n✅ Silver — AP por categoria")
print(ap.groupby("categoria")["valor_liquido"].sum().sort_values(ascending=False).apply(lambda x: f"R$ {x:,.0f}").to_string())
print("\n✅ Inadimplentes:", ar["fl_inadimplente"].sum(), "títulos")
```

```python
# ── Gold: Posição de caixa 30/60/90 dias ─────────────────────────────────────
ar_aberto = ar[ar["status"] != "Pago"].copy()
ap_aberto = ap[ap["status"] != "Pago"].copy()

entradas = ar_aberto.groupby("data_vencimento")["valor_liquido"].sum().reset_index()
entradas.columns = ["data", "entradas"]

saidas = ap_aberto.groupby("data_vencimento")["valor_liquido"].sum().reset_index()
saidas.columns = ["data", "saidas"]

posicao = pd.merge(entradas, saidas, on="data", how="outer").fillna(0).sort_values("data")
posicao["fluxo_liquido"]   = posicao["entradas"] - posicao["saidas"]
posicao["saldo_acumulado"] = posicao["fluxo_liquido"].cumsum()
posicao["horizonte"]       = posicao["data"].apply(lambda d:
    "30 dias"  if (d - hoje).days <= 30
    else "60 dias"  if (d - hoje).days <= 60
    else "90 dias"  if (d - hoje).days <= 90
    else "+90 dias")

resumo = posicao[posicao["data"] >= hoje].groupby("horizonte").agg(
    entradas=("entradas","sum"),
    saidas=("saidas","sum"),
    fluxo=("fluxo_liquido","sum"),
    saldo_min=("saldo_acumulado","min"),
    saldo_max=("saldo_acumulado","max"),
).reindex(["30 dias","60 dias","90 dias","+90 dias"])

print("✅ POSIÇÃO DE CAIXA — FUNDIÇÃO ABIFA")
print("="*55)
for h, row in resumo.iterrows():
    print(f"\n📅 {h}")
    print(f"   Entradas : R$ {row['entradas']:>12,.0f}")
    print(f"   Saídas   : R$ {row['saidas']:>12,.0f}")
    print(f"   Fluxo    : R$ {row['fluxo']:>12,.0f}  {'✅' if row['fluxo']>0 else '⚠️'}")
    print(f"   Saldo min: R$ {row['saldo_min']:>12,.0f}")

# Alertas
print("\n" + "="*55)
print("🚨 ALERTAS")
saidas_7d   = ap_aberto[ap_aberto["dias_vencimento"].between(0,7)]["valor_liquido"].sum()
entradas_7d = ar_aberto[ar_aberto["dias_vencimento"].between(0,7)]["valor_liquido"].sum()
if saidas_7d > entradas_7d * 1.5:
    print(f"⚠️  Saídas nos próximos 7d (R$ {saidas_7d:,.0f}) superam entradas (R$ {entradas_7d:,.0f})")
    print("   → Considerar antecipação de recebíveis de montadoras")

energia = ap_aberto[ap_aberto["categoria"]=="Energia Elétrica"]["valor_liquido"].sum()
print(f"⚡ Energia elétrica em aberto: R$ {energia:,.0f}")

saldo_minimo = posicao[posicao["data"] >= hoje]["saldo_acumulado"].min()
if saldo_minimo < 0:
    print(f"🔴 CRÍTICO: saldo projetado pode chegar a R$ {saldo_minimo:,.0f}")
else:
    print(f"✅ Saldo acumulado projetado permanece positivo (mín: R$ {saldo_minimo:,.0f})")
```

```python
# ── Gold: Modelo Prophet + Gráficos ──────────────────────────────────────────
serie = ext.groupby("data")["valor_sinalizado"].sum().reset_index()
serie.columns = ["ds", "y"]
serie["ds"] = pd.to_datetime(serie["ds"])

# Split treino/teste (hold-out 30 dias)
cutoff = serie["ds"].max() - pd.Timedelta(days=30)
treino = serie[serie["ds"] <= cutoff]
teste  = serie[serie["ds"] >  cutoff]

# Treinar modelo
modelo = Prophet(
    changepoint_prior_scale=0.05,
    seasonality_mode="additive",
    weekly_seasonality=True,
    yearly_seasonality=False,
    daily_seasonality=False
)
modelo.add_seasonality(name="mensal", period=30.5, fourier_order=5)
modelo.fit(treino)

# Avaliar no hold-out
prev_teste = modelo.predict(modelo.make_future_dataframe(periods=30))
prev_teste = prev_teste[prev_teste["ds"] > cutoff].set_index("ds")["yhat"]
real_teste = teste.set_index("ds")["y"]
idx = prev_teste.index.intersection(real_teste.index)
mape = mean_absolute_percentage_error(real_teste[idx], prev_teste[idx])
print(f"📊 MAPE no hold-out: {mape*100:.1f}%")

# Retreinar com todos os dados + prever 90 dias
modelo_final = Prophet(
    changepoint_prior_scale=0.05,
    seasonality_mode="additive",
    weekly_seasonality=True,
    yearly_seasonality=False,
    daily_seasonality=False
)
modelo_final.add_seasonality(name="mensal", period=30.5, fourier_order=5)
modelo_final.fit(serie)
futuro   = modelo_final.make_future_dataframe(periods=90)
forecast = modelo_final.predict(futuro)

# ── Gráfico 1: Histórico + Previsão ──────────────────────────────────────────
fig, axes = plt.subplots(3, 1, figsize=(14, 16))
fig.suptitle("Pipeline Fluxo de Caixa — Fundição ABIFA", fontsize=15, fontweight="bold", y=0.98)

hoje_ts = pd.Timestamp(hoje)
hist_plot = serie[serie["ds"] <= hoje_ts]
prev_plot = forecast[forecast["ds"] > hoje_ts]

ax = axes[0]
ax.plot(hist_plot["ds"], hist_plot["y"], color="#2563EB", linewidth=1.5, label="Histórico", alpha=0.9)
ax.plot(prev_plot["ds"], prev_plot["yhat"], color="#10B981", linewidth=2, linestyle="--", label="Previsão Prophet")
ax.fill_between(prev_plot["ds"], prev_plot["yhat_lower"], prev_plot["yhat_upper"],
                alpha=0.15, color="#10B981", label="IC 80%")
ax.axvline(x=hoje_ts, color="#EF4444", linestyle=":", linewidth=1.5, label="Hoje")
ax.axhline(y=0, color="#94A3B8", linestyle="-", linewidth=0.5)
ax.set_title(f"Fluxo de Caixa Diário — Histórico e Previsão 90 dias  |  MAPE: {mape*100:.1f}%", fontsize=12)
ax.set_ylabel("R$ Fluxo Diário")
ax.legend(fontsize=9)
ax.grid(alpha=0.3)
ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda x, _: f"R$ {x/1000:.0f}k"))

# ── Gráfico 2: Saldo acumulado projetado ─────────────────────────────────────
ax2 = axes[1]
saldo_prev = prev_plot["yhat"].cumsum().values
datas_prev = prev_plot["ds"].values
cor = ["#10B981" if s >= 0 else "#EF4444" for s in saldo_prev]
ax2.bar(datas_prev, saldo_prev, color=cor, alpha=0.7, width=1)
ax2.plot(datas_prev, saldo_prev, color="#7C3AED", linewidth=2)
ax2.axhline(y=0, color="#EF4444", linestyle="--", linewidth=1)
ax2.set_title("Saldo Acumulado Projetado — Próximos 90 dias", fontsize=12)
ax2.set_ylabel("R$ Saldo Acumulado")
ax2.grid(alpha=0.3)
ax2.yaxis.set_major_formatter(plt.FuncFormatter(lambda x, _: f"R$ {x/1000:.0f}k"))

# ── Gráfico 3: DRE por categoria (AP) ────────────────────────────────────────
ax3 = axes[2]
dre = ap.groupby("categoria")["valor_liquido"].sum().sort_values(ascending=True)
cores = ["#EF4444" if c in ["Energia Elétrica","Folha de Pagamento","Impostos"]
         else "#3B82F6" for c in dre.index]
bars = ax3.barh(dre.index, dre.values, color=cores, alpha=0.85)
ax3.set_title("Saídas por Categoria — Fundição (vermelho = alta prioridade)", fontsize=12)
ax3.set_xlabel("R$ Total")
ax3.xaxis.set_major_formatter(plt.FuncFormatter(lambda x, _: f"R$ {x/1000:.0f}k"))
for bar, val in zip(bars, dre.values):
    ax3.text(val + 5000, bar.get_y() + bar.get_height()/2,
             f"R$ {val/1000:.0f}k", va="center", fontsize=9)
ax3.grid(alpha=0.3, axis="x")

plt.tight_layout()
plt.savefig("fluxo_caixa_abifa.png", dpi=150, bbox_inches="tight")
plt.show()
print("✅ Gráficos gerados e salvos!")
```
